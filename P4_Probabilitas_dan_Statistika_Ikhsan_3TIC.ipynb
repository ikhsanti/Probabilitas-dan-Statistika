```python
import pandas as pd
dataraw = pd.read_excel("fulan.xlsx")
print(dataraw)
```

```python
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi")
print(datafrq)
```

```python
datafrq.plot()
```

```python
datafrq.plot(kind='bar')
```

```python
datafrq.plot(kind='pie', y='Frekuensi', autopct='%1.0f%%', legend=False)
```

```python
dataraw["Nilai"] = pd.to_numeric(dataraw["Nilai"], errors='coerce')
dt = dataraw["Nilai"]

stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()
print(stats)
```

```python
Q1 = dt.quantile(0.25)
Q3 = dt.quantile(0.75)
IQR = Q3 - Q1
batas_bawah = Q1 - 1.5 * IQR
batas_atas = Q3 + 1.5 * IQR

outlier = dt[(dt < batas_bawah) | (dt > batas_atas)]
print("IQR:", IQR)
print("Batas bawah:", batas_bawah, "| Batas atas:", batas_atas)
print("Outlier:", outlier.tolist())
```
